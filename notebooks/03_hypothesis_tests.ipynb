# 03. Hypothesis Tests

The tests compare item condition with top-ranked acceptance, distance with acceptance when coordinates exist, and urgency with time to confirmation. The unit of analysis is one top-ranked candidate per need. These are exploratory associations, not causal estimates; current generated results describe only the simulation.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display
from ml.analysis.hypothesis_tests import run_hypotheses

cwd = Path.cwd()
WORKSPACE = next((path for path in (cwd, *cwd.parents) if (path / "ml" / "analysis" / "hypothesis_tests.py").exists()), cwd)
report = run_hypotheses(WORKSPACE / "data" / "synthetic", expected_origin="synthetic")
for hypothesis, result in report["tests"].items():
    print(f"\n{hypothesis.replace('_', ' ').title()}")
    display(pd.Series(result))
print("Conclusion: p-values and effect sizes here describe only the synthetic generator. Re-run on consented pilot data before drawing community or operational conclusions.")
print("Assumptions/limitations: one top-ranked candidate per need reduces within-need duplication; needs are still assumed independent, and confirmed-only timing comparisons may be selection-biased.")
```
