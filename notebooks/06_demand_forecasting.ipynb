# 06. Demand Forecasting

This notebook uses a chronological holdout to compare LightGBM with a seasonal-naive baseline, then forecasts category demand with approximate residual intervals. No random split is used. A short synthetic history is only a pipeline demonstration and is not a planning forecast for real villages.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display
from ml.analysis.forecasting import forecast_demand

cwd = Path.cwd()
WORKSPACE = next((path for path in (cwd, *cwd.parents) if (path / "ml" / "analysis" / "forecasting.py").exists()), cwd)
needs = pd.read_csv(WORKSPACE / "data" / "synthetic" / "needs.csv")
try:
    result = forecast_demand(needs, forecast_months=3, validation_months=3, seed=42)
    print("Chronological validation metrics (synthetic):")
    display(pd.DataFrame(result["validation"]).T)
    print("Forecast with approximate prediction intervals:")
    display(result["future_forecast"])
    result["future_forecast"].pivot(index="month", columns="category", values="forecast").plot(marker="o", figsize=(11, 5))
    import matplotlib.pyplot as plt
    plt.title("Synthetic category demand forecast (LightGBM)")
    plt.ylabel("Need units (synthetic)")
    plt.tight_layout()
    plt.show()
    print("Conclusion: compare LightGBM with the seasonal-naive validation metrics before considering any category forecast. The short synthetic history is not a real village forecast.")
    print(f"Limitation: {result['limitation']}")
except (RuntimeError, ValueError) as error:
    print(f"Forecast not run: {error}")
    print("Conclusion: install requirements and provide at least six consecutive months for every category before forecasting. Do not infer unmet real demand from synthetic inputs.")
```
