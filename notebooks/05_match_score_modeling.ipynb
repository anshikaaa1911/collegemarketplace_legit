# 05. Match Scoring and Ranking

Compare the weighted-formula baseline, LightGBM regression, and LambdaRank using coordinator feedback. The holdout is grouped by need so candidates for one need cannot leak between training and evaluation. Metrics are RMSE against normalized feedback, NDCG@5, and Precision@3. Current synthetic feedback is generated for development only; it cannot establish real model quality.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display
from ml.matching.train import train_and_evaluate

cwd = Path.cwd()
WORKSPACE = next((path for path in (cwd, *cwd.parents) if (path / "ml" / "matching" / "train.py").exists()), cwd)
try:
    report = train_and_evaluate(WORKSPACE / "data" / "synthetic", expected_origin="synthetic", seed=42)
    metrics = pd.DataFrame(report["metrics"]).T
    display(metrics)
    print(f"Grouped split: {report['train_needs']} train needs / {report['test_needs']} test needs; {report['test_candidates']} held-out candidates.")
    print("Conclusion: choose a model only after it improves need-held-out ranking metrics on coordinator feedback. Synthetic outcomes are not evidence of real-world utility.")
    print(f"Limitation: {report['limitation']}")
except (RuntimeError, ValueError) as error:
    print(f"Model comparison not run: {error}")
    print("Conclusion: install requirements and collect coordinator decisions before treating any model as an evaluated recommender.")
```
